Al abrir este cuaderno se ve el último run que produjo los números del documento (D.1). Rutas relativas a la raíz del repo. El CSV se lee como `latin-1` porque mezcla `Sí`/`Si` en consentimiento.

```python
import csv, pathlib
raw = pathlib.Path("docs/mediciones/sus/sus-raw.csv").read_text(encoding="latin-1").splitlines()
rows = list(csv.DictReader(raw))
print(f"Participantes: {len(rows)}")
print("Codigos: " + ", ".join(r["codigo"] for r in rows))
scores = [float(r["sus_score"]) for r in rows]
print(f"Media verificada sus_score: {sum(scores) / len(scores):.1f}")
```

```python
import json, pathlib
est = json.loads(pathlib.Path("docs/mediciones/sus/estadisticas-sus.json").read_text(encoding="utf-8"))
print(f"n={est['n']} media={est['media']:.1f} DT={est['desviacion_tipica']:.3f} SE={est['error_estandar']:.4f} t(gl=14)={est['t_critico']:.4f}")
print(f"IC95=[{est['ic95_inf']:.2f}; {est['ic95_sup']:.2f}] adjetiva={est['adjetiva']}")
```
